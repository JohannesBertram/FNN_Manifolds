# Encoding/Decoding Manifold Explorer

Interactive 3-D viewer for neural manifolds built in `03_encoding_manifolds.ipynb`.

**Prerequisites:** Run `03_encoding_manifolds.ipynb` first to generate the IAN graph cache at `data/graphs/{PREFIX}_{R}.npz`.

**Four selection modes** (tabs below the scatter):

| Tab | How it works |
|-----|--------------|
| Click | Toggle individual points in/out of the selection |
| Cluster | Pick an HDBSCAN cluster from the dropdown |
| Bounding Box | Drag range-sliders to define an axis-aligned box |
| Radius | Click a seed point, then adjust the radius slider |

**PSTH panel:** ≤5 neurons → per-neuron strip; ≥6 neurons → population-averaged 3×2 grid.

**Live decoding panel:** Compares stimulus decodability of the selected subpopulation vs the full population.

```python
# ── Cell 3: Load from cache ───────────────────────────────────────────────────
try:
    data = load_for_explorer(PREFIX)
    print(f"Loaded  PREFIX={PREFIX}  R={data['R']}  "
          f"N_nonout={len(data['nonoutliers'])}  NSTIMS={data['NSTIMS']}  NDIRS={data['NDIRS']}")
except (FileNotFoundError, KeyError) as e:
    import glob as _glob
    _cached = _glob.glob(f'../data/graphs/{PREFIX}_*.npz')
    if not _cached:
        raise RuntimeError(
            f"IAN graph cache not found for PREFIX='{PREFIX}'.\n"
            f"Run 03_encoding_manifolds.ipynb first to generate the cache at "
            f"data/graphs/{{PREFIX}}_{{R}}.npz."
        ) from e
    raise
```

```python
# ── Cell 4: Launch explorer ───────────────────────────────────────────────────
explorer = ManifoldExplorer(
    embedding_      = data['embedding_'],
    tensor4d        = data['tensor4d'],
    nonoutliers     = data['nonoutliers'],
    neurons_used    = data['neurons_used'],
    categories      = data['my_stims'],
    NDIRS           = data['NDIRS'],
    extra_colorings = data['extra_colorings'],
)
explorer.show()
```
